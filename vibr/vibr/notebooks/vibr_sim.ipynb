```python
# Parameters for the plate
R = 0.1      # Radius (m)
h = 0.003    # Thickness (m)
cl = 0.01    # Mesh size (m)

# Geometry using pygmsh
with pygmsh.geo.Geometry() as geom:
    model = geom.__enter__()
    
    # Create a disk (circular plate)
    circle = model.add_cylinder([0, 0, 0], [0, 0, h], R, mesh_size=cl)
    
    # Generate and save the mesh
    mesh = geom.generate_mesh()
    mesh.write("circular_plate.vtk")  # for visualization
    mesh.write("circular_plate.vtu")  # for Sfepy
```
